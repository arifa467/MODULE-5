3. What are the different ways to define and call a function in Python?

  **Ans>>**  Python Functions is a block of statements that return the specific task. The idea is to put some commonly or repeatedly done tasks together and make a function so that instead of writing the same code again and again for different inputs, we can do the function calls to reuse code contained in it over and over again.

  Types of Functions-
  1. Built-in library function
  2. User-defined function

  **Define a Function**

  1. **The Standard (def) Keyword**: This is the most common and straightforward way to define a function in Python.

  2. **Anonymous Functions (Lambda Functions)**: These are small, single-expression functions that don't need a formal name defined with def. They are created using the lambda keyword.

  3. **Nested Functions (Functions Inside Functions)**: You can define a function within another function. The inner function has access to the variables in the outer function's scope (this is called a closure).

  **Calling a Function**

  - Using the function name: You call a function by its name followed by parentheses ().

  - **Passing arguments:** If the function definition includes parameters, you provide corresponding arguments inside the parentheses when calling the function. The order and type of arguments usually matter (unless you're using keyword arguments).

  - **Positional arguments**: Arguments are passed based on their position in the function call, matching the order of parameters in the definition (like in greet("Alice")).

  - **Keyword arguments**: You can call a function using the parameter names followed by an equals sign (=) to explicitly associate values with parameters. This allows you to pass arguments out of order and improves readability.

  - **Returning values:** Functions can optionally return values using the return statement. If a return statement is encountered, the function execution stops, and the specified value is sent back to the caller. If there's no return statement (or just return without a value), the function implicitly returns None.

5. What are iterators in Python and how do they differ from iterables?

  **Ans>>** **Iterables:** Think of an iterable as anything in Python that can be looped over. It's an object that is capable of returning its members one at a time. You can use them in a for loop.

  **Key characteristics of iterables:**
  - They have an __iter__() method. This method, when called, returns an iterator object for that iterable.
  - They don't necessarily store all their values in memory at once. For example, range(1000000) doesn't create a list of a million numbers; it just knows how to generate them on demand.

  Example:

  - Sequences: Lists ([]), tuples (()), strings (""), ranges (range())
  - Collections: Dictionaries ({}), sets ({})
  - File objects: Objects returned by open()

  **Iterators:** An iterator is an object that allows you to traverse through an iterable. It's like a cursor that points to an element in the iterable and knows how to move to the next one.

  **Key characteristics of iterators:**

  - Iterators remember their state – they know which element they are currently pointing to.
  
  - Once an iterator is exhausted (i.e., __next__() has been called until it raises StopIteration), you need to create a new iterator from the iterable to start iterating again from the beginning.






```python
# Example
lst = [1, 2, 3]       # Iterable
it = iter(lst)        # Iterator

print(next(it))
print(next(it))
```

6. Explain the concept of generators in Python and how they are defined.

  **Ans>>** A generator function is a special type of function that returns an iterator object. Instead of using return to send back a single value, generator functions use yield to produce a series of results over time. This allows the function to generate values and pause its execution after each yield, maintaining its state between iterations.

  **There are two main ways to define generators in Python:**

  1. Generator Functions: These are defined using the standard def keyword, just like regular functions, but they contain one or more "yield" statements.

  2. Generator Expressions: These provide an even more concise way to create simple generators using a syntax similar to list comprehensions but with parentheses () instead of square brackets [].

7. What are the advantages of using generators over regular functions?

  **Ans>>**  There are several advantages to using generators over regular loops. Here are a few of the most important ones:

  - **Memory Efficiency:** One of the biggest advantages of generators is that they are memory efficient. Since a generator yields its values one by one, it does not need to store all of the values in memory at once. This can be especially useful when working with large data sets, as it reduces the risk of running out of memory.

  - **Readability:** Generators can make your code easier to read and understand. By breaking up the iteration process into smaller chunks, you can write code that is more concise and easier to follow.

  - **Speed:** Generators can also be faster than regular loops, especially when working with large data sets. This is because they avoid the overhead of creating a list or other data structure to store the results.

  - **Statefulness:** Maintains state between calls using yield.

9. Explain the purpose and usage of the `map()` function in Python.

  **Ans>>** The map() function is used to apply a given function to every item of an iterable, such as a list or tuple, and returns a map object (which is an iterator).

  **Syntax of the map() function**:

  - map(function, iterable)

  **Purpose of map():** The primary goal of map() is to apply a given function to all elements of an iterable without explicitly writing a for loop. This can lead to more concise and often more readable code, especially for simple transformations. It embodies the idea of applying a consistent operation across a collection of items.

```python
# Example of map() Function

nums = [1, 2, 3]
squares = map(lambda x: x * x, nums)
print(list(squares))
```

```python
# 1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
# the list

def sum_even_num(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:
            total += num
    return total

sum_even_num([1,2,3,4,16,22,23])
```

```python
# 4. Write a Python function that checks if a given number is prime or not from 1 to 200

def is_prime(n):
    if n < 2 or n > 200:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

is_prime(89)
```

```python
# 5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
# terms.

class Fibonacci:
    def __init__(self, terms):
        self.terms = terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.terms:
            raise StopIteration

        result = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return result

fib = Fibonacci(10)  # create Fibonacci object for 10 terms

for num in fib:
    print(num)
```

```python
# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

for a in powers_of_two(5):
  print(a)
```

```python
# 7. Implement a generator function that reads a file line by line and yields each line as a string.

def read_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.rstrip('\n')  # Remove trailing newline characters

for line in read_file('example.txt'):
    print(line)
```

```python
# 8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

data = [(1, 3), (4, 1), (2, 2), (5, 0)]

sorted_data = sorted(data, key=lambda x: x[1])

print(sorted_data)
```

```python
# 9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
celsius_temperatures = [0, 20, 25, 30, 35, 40]
fahrenheit_temperatures = list(map(celsius_to_fahrenheit, celsius_temperatures))
print("Celsius temperatures:", celsius_temperatures)
print("Converted Fahrenheit temperatures:", fahrenheit_temperatures)
```

```python
# 10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

def not_vowel(char):
    return char.lower() not in 'aeiou'

input_string = str(input("Enter a String to Remove Vowel: "))

filtered_string = ''.join(filter(not_vowel, input_string))
print("Original string:", input_string)
print("String after removing vowels:", filtered_string)
```
